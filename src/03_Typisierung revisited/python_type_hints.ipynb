# Type Hint Syntax in Python

Referenz: Python 3.8+ (mit Hinweisen auf neuere Versionen)

## 1. Basis-Type-Hints

```python
x: int = 5
name: str = "Alice"
pi: float = 3.14
active: bool = True

count: int
```

## 2. Funktionen

```python
def add(a: int, b: int) -> int:
    return a + b

def log(msg: str) -> None:
    print(msg)
```

## 3. Standard-Container

```python
from typing import List, Dict, Tuple, Set

numbers: List[int] = [1, 2, 3]
mapping: Dict[str, int] = {"a": 1}
point: Tuple[int, int] = (3, 4)
unique: Set[str] = {"a", "b"}
```

### Ab Python 3.9 (empfohlen)

```python
numbers: list[int]
mapping: dict[str, int]
point: tuple[int, int]
```

## 4. Optional & Union

```python
from typing import Optional, Union

name: Optional[str] = None
value: Union[int, float] = 3.5

# Python 3.10+
name2: str | None = None
value2: int | float = 3.5
```

## 5. Any

```python
from typing import Any

data: Any = "anything"
```

## 6. Callable

```python
from typing import Callable

handler: Callable[[int, int], int]

def add(a: int, b: int) -> int:
    return a + b

handler = add
```

## 7. Iterable, Iterator, Generator

```python
from typing import Iterable, Iterator, Generator

def process(items: Iterable[int]) -> None:
    ...

def counter() -> Iterator[int]:
    yield 1

def gen() -> Generator[int, None, None]:
    yield 1
```

## 8. Eigene Klassen

```python
class User:
    def __init__(self, name: str) -> None:
        self.name: str = name
```

## 9. Forward References

```python
class Node:
    def __init__(self, next: "Node | None"):
        self.next = next
```

## 10. TypedDict

```python
from typing import TypedDict

class UserDict(TypedDict):
    id: int
    name: str

user: UserDict = {"id": 1, "name": "Alice"}
```

## 11. Literal

```python
from typing import Literal

status: Literal["open", "closed"] = "open"
```

## 12. Protocol (Duck Typing)

```python
from typing import Protocol

class HasClose(Protocol):
    def close(self) -> None: ...

def cleanup(obj: HasClose) -> None:
    obj.close()
```

## 13. Type Aliases

```python
UserId = int
Vector = list[float]

# Python 3.12+
# type UserId = int
```

## 14. Generics

```python
from typing import TypeVar

T = TypeVar("T")

def first(items: list[T]) -> T:
    return items[0]
```
