# Type Hints

## Syntax

* Jedes Objekt in Python ist eine Instanz einer Klasse
  * genauer: seit Python 3.0
* Die Instanzierung erfolgt entweder durch Aufruf des Konstruktors oder, falls vorhanden,  durch Verwendung eines Literals
* Der Typ kann durch die BuiltIn-Funktion `type` bestimmt werden

```python
names = ['A', 'B']
type(names)
```

```python
names = list()
names.append('A')
names.append('B')
type(names)
```

* Type Hints
  * `:`_type_ für Variable und Parameter
  * Rückgabewert mit `->`_type_ 
  * Für Collections _collectiontype_`[`_elementtype_`]`
    * gesprochen _Collection of_, z.B. `list[str], eine _List of String_ 

```python
name :str = 'Hugo'
names :list[str] = ['A', 'B']
```

## Motivation zum Einsatz von Type Hints

* Type Hints werden zur Laufzeit **nicht** geprüft
* Sie dienen damit nur als Metainformationen für andere Tools
  * Autovervollständigung in einer IDE wie Visual Studio Code
  * Statische Typprüfungen durch Linter
    * englisch 'lint' = Fussel, ein Linter ist damit ein 'Entfusseler'
* Type Hints für Variablen sind unüblich, da durch ausgefeilte Type-Inference überflüssig

```python
number :int = int('42') # was denn sonst???
```

* Python ist laut Umsetzung keine statisch typisierte Programmiersprache, deshalb ist eine Prüfung zur Laufzeit natürlich theoretisch möglich, aber immer aufwändig
  * z.B. Modul `typeguard`
