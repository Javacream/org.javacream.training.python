Dieses Notebook enthält eine kurze, allgemeine Einführung in **REST** und erzeugt/zeigt einen **RESTful WebService** (Flask) für:

`Person(lastname, firstname, height, weight)`


## REST – die Kernideen
- **Ressourcen** (Nomen) statt Aktionen (Verben) im URL-Pfad: `/persons`, `/persons/1`
- **HTTP-Methoden** drücken die Aktion aus: `GET`, `POST`, `PUT`, `PATCH`, `DELETE`
- **Stateless**: jeder Request ist in sich vollständig
- **Repräsentationen**: JSON ist Standard
- **Statuscodes** sind Teil der Semantik (`201`, `404`, `400`, ...)


## CRUD-Mapping (typisch)
| Zweck | Methode | Endpoint |
|---|---|---|
| Liste lesen | GET | `/persons` |
| Objekt erstellen | POST | `/persons` |
| Objekt lesen | GET | `/persons/<id>` |
| Objekt ersetzen | PUT | `/persons/<id>` |
| Objekt teilweise ändern | PATCH | `/persons/<id>` |
| Objekt löschen | DELETE | `/persons/<id>` |


## Statuscodes (Quick Reference)
- **200 OK**: Erfolg (GET/PUT/PATCH)
- **201 Created**: neu erstellt (POST), häufig mit `Location` Header
- **204 No Content**: Erfolg ohne Body (DELETE)
- **400 Bad Request**: ungültige Eingabe
- **404 Not Found**: Ressource nicht vorhanden
