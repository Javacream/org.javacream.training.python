## Was ist Tkinter?
**Tkinter** ist das mit Python ausgelieferte GUI-Framework und ein Wrapper um die Tcl/Tk-Bibliothek.

Eigenschaften:
- Bestandteil der Python-Standardbibliothek
- Plattformübergreifend (Windows, macOS, Linux)
- Ereignisgesteuertes Programmiermodell
- Gut geeignet für kleine bis mittlere Desktop-Anwendungen

Nicht-Ziele:
- Moderne Web-Optik
- Hochperformante 3D-GUIs


## Grundstruktur einer Tkinter-Anwendung
Eine typische Tkinter-App besteht aus:

1. Erzeugen des Hauptfensters (`Tk`)
2. Erzeugen von Widgets (Label, Button, Entry, …)
3. Anordnen der Widgets (Layout)
4. Starten der Event-Schleife (`mainloop()`)


```python
import tkinter as tk

root = tk.Tk()              # Hauptfenster
root.title("Meine erste Tkinter-App")

label = tk.Label(root, text="Hallo Tkinter!")
label.pack()

root.mainloop()             # Event-Schleife
```

## Widgets
Widgets sind die Bausteine einer GUI.

Häufige Widgets:
- `Label` – Text anzeigen
- `Button` – klickbarer Knopf
- `Entry` – einzeilige Texteingabe
- `Text` – mehrzeiliger Text
- `Checkbutton` – Checkbox
- `Radiobutton` – Optionsauswahl
- `Frame` – Container für andere Widgets


```python
import tkinter as tk

root = tk.Tk()
root.title("Widgets")

tk.Label(root, text="Name:").pack()
entry = tk.Entry(root)
entry.pack()

def say_hello():
    print("Hallo", entry.get())

tk.Button(root, text="Klick mich", command=say_hello).pack()

root.mainloop()
```

## Layout-Manager
Tkinter kennt drei Layout-Manager:

### `pack()`
- Einfach, stapelt Widgets
- Gut für kleine GUIs

### `grid()`
- Tabellenlayout (Zeilen/Spalten)
- Am häufigsten genutzt

### `place()`
- Absolute Positionierung
- Selten empfohlen

**Wichtig:** In einem Container (z. B. `Frame`) nur **einen** Layout-Manager verwenden.


```python
import tkinter as tk

root = tk.Tk()
root.title("grid() Beispiel")

tk.Label(root, text="Vorname").grid(row=0, column=0, sticky="e")
tk.Label(root, text="Nachname").grid(row=1, column=0, sticky="e")

tk.Entry(root).grid(row=0, column=1)
tk.Entry(root).grid(row=1, column=1)

tk.Button(root, text="OK").grid(row=2, column=0, columnspan=2)

root.mainloop()
```

## Events und Callbacks
Tkinter ist **ereignisgesteuert**:
- Klicks
- Tastatureingaben
- Mausbewegungen

Zwei Varianten:
- `command=` (einfach)
- `bind()` (flexibel)


```python
import tkinter as tk

root = tk.Tk()
root.title("Events")

def on_click():
    print("Button geklickt")

def on_key(event):
    print("Taste:", event.keysym)

tk.Button(root, text="Klick", command=on_click).pack()
root.bind("<Key>", on_key)

root.mainloop()
```

## Variablenklassen
Tkinter bietet spezielle Variablen, die Widgets automatisch bei Änderungen aktualisieren:

- `StringVar`
- `IntVar`
- `DoubleVar`
- `BooleanVar`

Sehr praktisch für Formulare.


```python
import tkinter as tk

root = tk.Tk()

name = tk.StringVar(value="Max")

tk.Entry(root, textvariable=name).pack()

def show():
    print("Name:", name.get())

tk.Button(root, text="Auslesen", command=show).pack()

root.mainloop()
```

## Frames
Größere GUIs sollten in **Frames** strukturiert werden

Vorteile:
- Bessere Lesbarkeit
- Eigene Layout-Manager pro Frame
- Wiederverwendbare Komponenten


```python
import tkinter as tk

root = tk.Tk()

top = tk.Frame(root, bg="lightgray")
top.pack(fill="x")

bottom = tk.Frame(root)
bottom.pack(fill="both", expand=True)

tk.Label(top, text="Kopfbereich").pack(padx=10, pady=10)
tk.Button(bottom, text="Aktion").pack(pady=20)

root.mainloop()
```

## Objektorientierter Stil
Tkinter lässt sich sehr gut objektorientiert nutzen


```python
import tkinter as tk

class App(tk.Tk):
    def __init__(self):
        super().__init__()
        self.title("OO Tkinter App")
        self._build()

    def _build(self):
        tk.Label(self, text="Hallo OO-Welt").pack()
        tk.Button(self, text="Beenden", command=self.destroy).pack()

app = App()
app.mainloop()
```
