## Cleanup

* Droppen der überflüssigen Spalte

```python
df.drop(columns=['Unnamed: 0'])
```

```python
df.head()
```

* Vorsicht
    * Operationen, die die Struktur eines DataFrames ändern, lassen das Original unverändert
        * Sie geben einen neuen DataFrame zurück

```python
cleaned_df = df.drop(columns=['Unnamed: 0'])
cleaned_df
```

* inplace=True wird das Original verändern

```python
unnecessary_column_name = 'Unnamed: 0'
if unnecessary_column_name in df.columns:
    df.drop(columns=[unnecessary_column_name], inplace=True)
df.head()
```
