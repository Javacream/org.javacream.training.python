
# Analyse des Diamanten-Datensatzes

* Die Daten stammen aus einer Mine und erfassen die geschürften Diamante aus dem Frühjahr 2024 

## Fachliche Informationen

* carat: Weight of the diamond
* depth: Total depth percentage (2 * z / (x + y))
* table: Width of the top facet relative to the widest point.
* x: Length in mm
* y: Width in mm
* z: Depth in mm
* price: Price in US dollars
* cut: Quality of the cut
* color: Color grade from D (best, colorless) to J (worst, faint yellow/brown)
* clarity: Measurement of internal flaws (I1 (worst) to IF (best))
