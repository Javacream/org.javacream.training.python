# BMI Berechung und Kategorisierung

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('people.csv')
```

```python
df.info()
```

## Bereinigung der NaN

```python
df['Height'] = df['Height'].fillna(183)
df['Weight'] = df['Weight'].fillna(81.8)
df.info()
```

## BMI-Berechnung

```python
df['Bmi'] = df['Weight']/((df['Height']/100)**2)
df.info()
```

## BMI Kategorisierung

```python
df['BMI Category'] = ''
df.loc[df['Bmi'] < 18.5, 'BMI Category'] = 'underweight'
df.loc[(df['Bmi'] >= 18.5) & (df['Bmi'] < 25), 'BMI Category'] = 'normalweight'
df.loc[(df['Bmi'] >= 25) & (df['Bmi'] < 35), 'BMI Category'] = 'overweight'
df.loc[df['Bmi'] >= 35, 'BMI Category'] = 'obese'
df
```
