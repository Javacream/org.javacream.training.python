## Der DataFrame

```python
df = pd.read_csv('people.csv')
df.head(3) # default = 5
```

### Hinweis
* Jeder DataFrame hat einen Index
* Dieser Index identifiziert eine Zeile des DataFrames
* Im Standard eine fortlaufende Nummer
* Die Spalten werden ebenfalls identifizierbar gemacht über eindeutige Spaltennamen
* Als Index kann aber auch eine andere Spalte gesetzt werden
    * Zugriff auf eine Spalte erfolgt wie bei einem Dictionary über den Namen der Spalte

```python
df.set_index('Name')
```

### Hinweis
* Viele Methoden eines DataFrames scheinen den DataFrame zu verändern
* Das ist aber nicht der Fall, es wird im Standard immer ein neuer DataFrame zurückgegeben, der Original-DataFrame bleibt unverändert
  * Es gibt einen Parameter inplace. Falls dieser auf True gesetzt ist, wird der Original-DataFrame verändert
    * Vorsicht, damit kann es passieren, dass Zellen nicht beliebig oft ausgeführt werden können

```python
df
```

```python
df2 = df.set_index('Name')
df2
```

```python
df2.rename(columns={'Weight': 'PERSON_WEIGHT'}, inplace=True)
df2
```

```python
df['Answer'] = 42
df.head()
```

```python
df.loc[0, 'Name'] = 'CHANGED' # df[repr((0, 'Name))]
df.head()
```

```python
df.iloc[1,2] = 4711
df.head()
```

```python
df2 = df.set_index('Name')
df2.head()
```
