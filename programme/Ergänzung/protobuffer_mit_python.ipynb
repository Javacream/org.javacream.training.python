## Allgemein

* [Wiki-Artikel](https://de.wikipedia.org/wiki/Protocol_Buffers)

## Grundidee

* Sprachübergreifendes Format zur Definition von Datenstrukturen
* Generierung von Code, der für ausgewählte Programmiersprachen die Datenstruktur umsetzt
* Encoding / Decoding in ein kompakt optimiertes Format zur effizienten Übertragung über Netzwerk


### Definition der Datenstruktur

* Datei person.proto

    syntax = "proto3";    
    
    message Person {    
        string name = 1;    
        int32 id = 2;    
        string email = 3;     
    }    

### Python

* Native Installation des protoc-Tools, siehe https://protobuf.dev/installation/

* Erzeugen des Moduls

protoc --python_out=. person.proto

#### Ein simples Programm

Hinweis: Protobuf-Dateien werden normalerweise z.B. über einen RESTService über das Netz versendet, nicht lokal abgelegt...

* Schreiben der Protobuf-Datei

```python
import person_pb2

# Create a new person
person = person_pb2.Person()
person.name = "Alice"
person.id = 1234
person.email = "alice@example.com"

# Serialize to binary
binary_data = person.SerializeToString()

with open('person.pb', 'wb') as file:
    file.write(binary_data)

person
```

* Lesen der Protobuf-Datei

```python
with open('person.pb', 'rb') as file:
    binary_data = file.read()

# Deserialize from binary
new_person = person_pb2.Person()
new_person.ParseFromString(binary_data)

print(f"Name: {new_person.name}")
print(f"ID: {new_person.id}")
print(f"Email: {new_person.email}")
```
