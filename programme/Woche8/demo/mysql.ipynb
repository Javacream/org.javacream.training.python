# Zugriff auf eine MySql-Datenbank

* Installation des Treibers mit `pip install mysql-connector-python`

* import des Connectors

```python
import mysql.connector
```

* Aufbau der Verbindung zur externen Datenbank

```python
connection = mysql.connector.connect(
        host='javacream.eu', port=3406, user='user', password='user', database='javacream'
    )
```

* Öffnen des Cursors

```python
cursor = connection.cursor()
```

* Absetzen des Statements

```python
cursor.execute("SELECT * FROM BOOKS")
```

Fetchen und Ausgeben der Ergebnisse

```python
result = cursor.fetchall() 
for row in result:
    print(row)
```
