## Maximal- und Minimalpreis der Bücher

```python
cursor = conn.cursor()
cursor.execute('select max(price) from book')
cursor.fetchone()
```

```python
cursor = conn.cursor()
cursor.execute('select min(price) from book')
cursor.fetchone()
```

## Anzahl der Bücher eines Autoren

```python
cursor = conn.cursor()
cursor.execute('select count(*) from book_author where author_id = 4')
cursor.fetchone()
```
