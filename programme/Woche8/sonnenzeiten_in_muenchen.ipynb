# Sonnenzeiten in München

```python
import csv
from datetime import datetime

DATE_FORMAT = "%H:%M"

with open('Sonnenzeiten_München_2025.csv', encoding='utf-8') as file:
    reader = csv.reader(file)
    next(reader)
    data = [(row[0], (datetime.strptime(row[1], DATE_FORMAT)), (datetime.strptime(row[2], DATE_FORMAT))) for row in reader] 
```

## Erstellen eines Images

* Hinweis: plt ist quasi eine leere, bemalbare Zeichenfläche, ein Canvas

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 6))
plt.title('Sonnenauf- und -untergang in München')
x, y1, y2 = zip(*data)
y1 = [t.hour * 60 + t.minute for t in y1]
y2 = [t.hour * 60 + t.minute for t in y2]
plt.plot(x, y1, label='Sonnenaufgang')
plt.plot(x, y2, label='Sonnenuntergang')
plt.legend()
plt.show()
```
