# Allgemein

* Erste Näherung
    * Ein DataFrame ist nichts anderes als eine Tabelle
    * Damit ist eine Sicht auf die Daten ist eine "Liste von Listen"
* Genauer
    * Ein DataFrame besteht aus Series
    * Eine Series besteht aus einer Index-Spalte und einer Wertspalte
        * Damit ist eine Series bereits selber eine Minitabelle

```python
import pandas as pd

names_list = ['Hugo', 'Hanna', 'Andrea'] # ganz normale Python-Liste
print(names_list)
names_series = pd.Series(names_list)
print(names_series)
```

* Series mit Werte- und Index-Liste

```python
names_series = pd.Series(names_list, index= ['a', 'b', 'c'])
print(names_series)
```

* Series aus einem Dictionary

```python
names_dict= {'a': 'Hugo', 'b': 'Hannah', 'c': 'Andrea'}
names_series = pd.Series(names_dict)
names_series
```

### Series-Operationen
