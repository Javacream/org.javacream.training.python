# Berechnung des BMI für die Personen der people.csv

* Hinweis: Nicht-berechenbare BMIs sind 'NaN'

```python
import pandas as pd

df = pd.read_csv('people.csv')
df['bmi'] = df['weight_kg']/(df['height_cm'] * df['height_cm']) * 10000
df['result'] = df['firstname'] + " " + df['lastname'] + "->" + df['bmi'].astype(str)
df['result'].to_csv('bmi.csv')
df
```
