* Weisen Sie allen nicht-gesetzten Körpergrößen den Wert 181 zu, allen nicht gesetzten Körpergewichtete 77.7
* Einführung einer klassifizierenden Spalte 'height_category'
  * größer 190 -> 'tall'
  * 175 - 190 -> 'normal'
  * kleiner 175 -> small


```python
df = pd.read_csv('people.csv')
df.fillna({'height_cm': 181}, inplace=True)
df.fillna({'weight_kg': 77.7}, inplace=True)
df['height_category'] = 'normal'
df.loc[df['height_cm'] > 190, 'height_category'] = 'tall'
df.loc[df['height_cm'] < 175, 'height_category'] = 'small'
df
```
