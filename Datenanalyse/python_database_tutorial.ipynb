## Konkretes Beispiel: Zugriff auf eine embedded sqlite

* Importieren des sqlite3-Moduls

```python
import sqlite3
```

* Aufbau von Connection und Cursor
  * Hinweis: :memory: startet eine Sqlite-Datenbank innerhalb des Python-Prozesses

```python
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()
```

* Anlegen einer Testdatenbank

```python
sql_statement = 'create table messages (message varchar(128))'
cursor.execute(sql_statement)
```

* Einfügen von Testdaten

```python
cursor.execute("insert into messages values ('Hello')")
cursor.execute("insert into messages values ('World')")
```

* Auslesen der Daten

```python
cursor.execute("select * from messages")
result = cursor.fetchall()
for row in result:
    print(row)
```

* Schließen von Cursor und Connection

```python
cursor.close()
conn.close()
```
