# Ein Datensatz mit Büchern

## Einlesen der Daten

```python
import pandas as pd

df = pd.read_json('https://raw.githubusercontent.com/ozlerhakan/mongodb-json-files/master/datasets/books.json', lines=True)
df.set_index('_id', inplace=True)
df.drop(['publishedDate', 'longDescription', 'status', 'thumbnailUrl'], inplace=True, axis=1)
df.head()
```

* Hinweis
  * Die JSON-Datei enthält Book-Objekte zeilenweise, nicht in Form einer Liste
  * Deshalb ist beim Einlesen der Parameter lines=True zu setzen

## Beschreibung

## Dicke Bücher, die mehr als 800 Seiten haben

```python
df[df['pageCount'] > 800]
```

## Die 25% der seitenstärksten Bücher

```python
quantile_75 = df['pageCount'].quantile(0.75)
df[df['pageCount'] > quantile_75]
```
