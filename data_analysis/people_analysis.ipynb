# Analyse des Personen-Datensatzes

## Datensatz

* PEOPLE.csv
* Abzug aus der Unternehmens-Datenbank

## Analyse

### Einlesen der Daten

```python
import csv
people = []
with open('PEOPLE.csv', encoding='utf-8') as file:
    reader = csv.DictReader(file)
    for row in reader:
        row["height"] = float(row["height"])
        row["weight"] = float(row["weight"])
        people.append(row)
```

## Was für Spalten haben wir? Was sind die Datentypen?

```python
person = people[0] 
print(person.keys())
for attribute in person.keys():
    print(f'{attribute}: {type(person[attribute])}')
```

## Wie viele Datensätze haben wir?

```python
len(people)
```

## Was ist das maximale Körpergeweicht? Wer hat dieses Gewicht

```python

max_weight = max([person['weight'] for person in people])
max_weight
```

```python
[person for person in people if person['weight'] == max_weight]
```

## Welche Personen mit einem Gewicht kleiner 75 haben einen Nachnamen, der mit 'S' beginnt?

```python
[person for person in people if person['weight'] < 75 and person['lastname'].startswith('S')]
```
