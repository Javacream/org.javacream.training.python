### DataFrame als Datencontainer / Collection

* Elementare Berechnungen / Zugriffe

```python
height_series = df['height'] + 1
```

```python
df['weight']/(df['height']**2)
```

```python
bmi = df['weight']/(df['height']**2)
bmi
```

```python
bmi.info()
```

```python
bmi.describe()
```

* Zuweisung einer Series an einen DataFrame

```python
df['bmi'] = df['weight']/(df['height']**2)
df.info()
```

* Entfernen einer Series aus einem DataFrame

```python
df.drop('GENDER', axis=1)
df.drop(columns=['GENDER'])
```

* Detail: inplace-Operationen

```python
dropped = df.drop(columns=['GENDER'])
dropped
```

```python
df.drop(columns=['GENDER'], inplace=True)
df
```

### Indexzugriff

```python
df.index[0]
```

```python
print(type(df.loc))
sliced_df = df.loc[0:2, ['height', 'weight']]
sliced_df.describe()
```
